
## Reinforcement Learning
__Reinfrocemnt Learning__ is a type of machine learning tasks where decisions makers are trying to maximize long-term rewards or minimize long-term costs. In the leanring proces, the decision-makers learn how to maximize rewards or minimize costs throught the interaction expereinces in the enviroment (key feature  is guided by reward signals or cost signals).

_example_ robot in a maze observes its surroundigns and treis to move, stupid moves waste time and enegy while smart moves lead it out of the maze. in this example ove is the action after its obersvation and time and enrgy sit spends is the cost (negative rewards) 




### key terms

<ls>

 1. __Reward__ (or __cost__): the goal of RL is to maximize the long-term reward, or minimize the long-term cost. in the example of the maze, the time and energy wasted  gereante negative rewards. The robot wants to get postiive rewards while avoiding ngeative rewards.

 2. __Policy__ : Decision-maker will act according to their obsercations, and the way of making decisions is called _policy_.  It' is the policy that RL algorithm learns. RL algorithm changes policy to maximize the total rewards.

### Agent-Enviroment Interface
**Agent-enviroment interface** is a tool to analyze an RL system
1. __Agent__ is the decision-maker and learning. As __decision-maker__ the agent can decide what action to take. As __learner__ the agent can adjust its policty. There can be one agent or many agents in an RL system
2. __Enviroment__ inclues everytthign in RL system that agents do not include. it is what the aents interacts with
  
_example_ At work i can decide what to do, and leanrr to adjust polcy that determins what to do. Those belong to the *agent*. Meanwhile my health, tiredness and starvatyion are things i can not control directly. so they belon gto the *enviroment*. The agent can respond to the enviroment. Say i may go to bed if i feel tired,

_note_ while  RL systmes are allyzed using agent-enviroment interface, in most cases it is not imperativie to use agent-enviroment-interfaces.
* Agent observe teh enviroment and get the __observation__ denoted *O*
* Agent use some polcy to determin the __action__ denoted *A*
* after being impacted by the action, the enviroment signals _rewards__ dennoted _R_ and cahange its __state__ denoted *S*

_note_ States, observations and actions are not necessarily nmeric values, they can be stuff like feeling hungry and eating (denoted in sans-serif). Rewards are always numerical values (scalers denoted in serif)

_note_ most RL systems are casual which means what happened eariler can only impact what happens aftewards. Sequential decision makign  can be indexed on time denote _t_, the state at time _t_ can be denoted as $S_{t}$. Independent events do not require a time elemetn, such as rolling a dice and the reward is the outcome of the dice. 

when the intereaction timing between an agent and enviroment is countable it is called a discrete-time agent-enviroment interface. the intearction timing index can be mapped to $t = 1,2,3,...$

_note_ some liteature denote teh reard signal after teh action $A_t$ as $R_t$ while others might denote it as $R_{t+1}$ (<- will used $t+1$ moving on)



### Taxonomy of RL
There are various ways to classify RL based on tasks, or based on algorithms

- single/multi-agent tasks (__MALR__): in a 
__single-agent tasks__ there is only one agent who can observe, in multi-agent, each agent may just osberse its own obsercation. in multi-agent  taks agents may or may not communicate. Each agent may have differnt goals that might conflict.
- episodic/continous tasks : an __episodic tasks__ has start and end. For example a game of Go  is an episode tasks where the board is empty at the beginning and full at the end. a __continous tasks__ has neither beginning nor end.  for example: te scheduler of the public cloud servies needs to process without interruption forever
- discrete/continious-time tasks : __discrete-time tasks__ : time index is countably infinite or the time index is finite wher e$T$ can be a random variable. __continious-time__ if the time interaction between agent and the enviroment is uncountalbe i.e. all non-negative real numbers or non-empty coutnious subset. __note__ classifcation is not excluseive, additonally the intervals between two interactions can be stochastic and the inteval can futher impact sequetional evolutions.
- discrete/continious action space: __discrete action space__ - coutnable outcomes (only moving north,east, south, west). __continous actions space__ - uncountable outcome (moving between $[ 0\degree,  360\degree ]$)
- deterministic/ stochastic enviroment: __deterministic enviroment__ - the enviroment does not change, for example a robot in a fixed maze. __stochastic enviroment__ - the enviroment is stochastic. for example the robot is in a randomally generated maze
- fully/partially observalbe enviroment: __fully observable enviroment__ - agent can see the whole enviroment. for example in the game of Go because players can see the whole board. __partially obsevable__ agents can onlys ee part of the enviroment. for example in poker, players do not know what cards the opponets have
