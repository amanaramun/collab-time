### Taxonomy of RL
There are various ways to classify RL based on tasks, or based on algorithms


#### Task-based


- single/multi-agent tasks (__MALR__): in a 
__single-agent tasks__ there is only one agent who can observe, in multi-agent, each agent may just osberse its own obsercation. in multi-agent  taks agents may or may not communicate. Each agent may have differnt goals that might conflict.
- episodic/continous tasks : an __episodic tasks__ has start and end. For example a game of Go  is an episode tasks where the board is empty at the beginning and full at the end. a __continous tasks__ has neither beginning nor end.  for example: te scheduler of the public cloud servies needs to process without interruption forever
- discrete/continious-time tasks : __discrete-time tasks__ : time index is countably infinite or the time index is finite wher e$T$ can be a random variable. __continious-time__ if the time interaction between agent and the enviroment is uncountalbe i.e. all non-negative real numbers or non-empty coutnious subset. __note__ classifcation is not excluseive, additonally the intervals between two interactions can be stochastic and the inteval can futher impact sequetional evolutions.
- discrete/continious action space: __discrete action space__ - coutnable outcomes (only moving north,east, south, west). __continous actions space__ - uncountable outcome (moving between $[ 0\degree,  360\degree ]$)
- deterministic/ stochastic enviroment: __deterministic enviroment__ - the enviroment does not change, for example a robot in a fixed maze. __stochastic enviroment__ - the enviroment is stochastic. for example the robot is in a randomally generated maze
- fully/partially observalbe enviroment: __fully observable enviroment__ - agent can see the whole enviroment. for example in the game of Go because players can see the whole board. __partially obsevable__ agents can onlys ee part of the enviroment. for example in poker, players do not know what cards the opponets have


#### Algorithem-based

- on/off policy learning: __on-policy__ - learns from the policy that is ued to play (example Imagine you're trying to teach a robot to navigate a maze. In on-policy learning, the robot learns based on the actions it is currently taking. It's like learning to cook by trying out different recipes yourself.) __off-policy__ - can lean from records of games with old polcies ( learning from someone else's experience. In this approach, the robot might watch another robot navigate the maze and learn from its actions. - model- based/free learning : __Model-based__ reinforcement learning uses a model of the environment to simulate outcomes and plan actions, while __model-free__ reinforcement learning learns directly from interactions with the environment without building a model. Model-based methods can be more efficient but are often more complex to implement than model-free methods. example model-based game of Go the rules of the gam  are the model and we know them prior.

- deep/ classical reinforcment learning:  if an algorithm use neral networks as a componet, it leverages __deep learning (DL)__ and is called __deep RL__. example video game AI can use convlutional neural networks to process video game images and solve RL taks, since it uses both RL and neural networks it is a __Deep Reinforcment learning algorithm__

### Performance Metrics

* __Long-tern Reward__: this indicates the long-term reward that an algorithm can achieve when it is applied to the task. Many statistics are derived from long-term rewards such as mean and st.dev. We usually want larger returns and episode rewards. We also may want small Std.devs
* __optimality and convergence__: we perfer the algorithms that can converage to the optimal
* __sample complexity__ :  it is usually costly and time-consuming to interact with enviroments we we refer to learn from as few samples as possible. Somealgorithms can make use of the enviroment models or historical records to learn so they have smaller sample compleixty. 

>_note_ Some model-based algorithms can even solve the tasts wihtout learning from samples. strictly speaking,s uch algorithms are not RL algorithms. Algorithms for offline tasks learn from historical records, they do not interact with the enviroments. Therefore the "online" sample complxity of these offline RL is 0, however offline RL tasks usually get a fixed number of samples so the offline sample complexity is critical.

* __regeret__: regert is defined as the summation of episode regerts during the training process. __Regret__ is define das the difference between teh acatual episode reweard and the optimal episode reward after the training completes. we prefer small regert values. Some RL algorithms learn from scratct and accumlate regert , others are pretrained in one cheaper enviroment then trained in a costly enviroment. We only care about the regert in the costly enviroment. example: an agent can be trained in a simulation enviroment before it is formally trianed in the real workd becuase failure in the real world is costly. 

* __scalability__: some RL algorithms can use multipleworkers to accelerate learning.Since RL tasks are usually time-consuming an dcomputation-intensive, it is an advauntage to ahve scalability to sepped up learning.

* __convergence speed, training time and complexity__: Theorist care whether an RL algorithm converages and how fast it converges. Time complexity can also be used to indicate the speed of convergence. Convergence is a good property and we prefer small time complexity

> _note_ Convergence in reinforcement learning (RL) refers to the process where an RL algorithm's performance stabilizes, meaning that the agent's learned policy consistently yields optimal or near-optimal rewards over time. This typically occurs when the agent has sufficiently explored the environment and refined its strategy based on the feedback received

* __occupied space and space complexity__ : we perfer small space occupation, some use cases particualy care about the space for storing historical records while other use cases care more about the storeage on GRAM

### Closed-form agent for tasks 
next we prepare an agent for interacting with the enviroment. Genreally, gym does not inlcue agents so we need to implement an agent ourselve. the class `ClosedFormAgent` will be the agent. it is a simplifed agent that only decides but can not learn. THerefore exactaly speaking this is not an RL agent, but it is enough to show interaction between agent and enviroment



````python
class ClosedFormAgent:
    def __init__(self, _):
        pass

    def reset(self, mode=None):
        pass

    def step(self, observation, reward, terminated):
        position, velocity = observation
        lb = min(-0.09 * (position + 0.25) ** 2 + 0.03,
                0.3 * (position + 0.9) ** 4 - 0.008)
        ub = -0.07 * (position + 0.38) ** 2 + 0.07
        if lb < velocity < ub:
            action = 2 # push right
        else:
            action = 0 # push left
        return action

    def close(self):
        pass

````

Now there is an agent and an enviroment to interact with each other. We now need to make something to place the episode

* `env`: is the enviroment object
* `agent`: is an agent object
* `seed` : `None` or an `Int` which will be used as the seed of a random number generator
* `mode`: `None` or "train". if the string is "train" it tires to tell teh agen tto learn from interactions, hwoever if the agent can not learn, it is no use
* `render`: `bool` value indcating wheter to visulize the interaction or not

this function returnes `episode_reward` and `elapses_step`. `episode_reward` is a `float` that indicates reward and `elapse_step` is an `Int` denote the number of steps in an episode

````python
def play_episode(env,agent, seed = None, mode = None, render = False):
    observation, _ = env.reset(seed = seed)
    reward, terminated, truncated = 0. , False, False
    agent.reset(mode = mode)
    episode_reward, elapsed_steps = 0., 0
    while True:
        action = agent.step(observation, reward, terminated)
        if render:
            env.render()
        if terminated or truncated:
            break
        observation, reward, terminated, truncated,_ = env.step(action)
        episode_reward += reward
        elapsed_steps += 1
    agent.close()
    return episode_reward, elapsed_steps
````

putting all three together `env = gym.make('MountainCar-v0')` , `CloseFormAgent`, and `play_episode()`we have an visulization

````python
env.step(0)
````

````python
env = gym.make('MountainCar-v0')
agent = ClosedFormAgent(env)

logging.info('==== test ====')
episode_rewards = []
for episode in range(100):
    episode_reward, elapsed_steps = play_episode(env, agent)
    episode_rewards.append(episode_reward)
    logging.info('test episode %d: reward = %.2f, steps = %d',
            episode, episode_reward, elapsed_steps)
logging.info('average episode reward = %.2f ± %.2f',
        np.mean(episode_rewards), np.std(episode_rewards))

env.close()
````
