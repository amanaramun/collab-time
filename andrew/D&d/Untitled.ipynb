```python
import pandas as pd
```

```python
disc_dic = {}
```

```python
data_sources = ['Basic', 'GEONAICS']
sectors = ['22','23']
files = ['c','d']
```

```python
for source in data_sources:
    disc_dic[source] = {}
    for sector in sectors:
        if source == 'Basic' and sector in ['23','31','23']:
            print(sector, 'mmc')
            disc_dic[source][sector]= {}
            for file in files:
                disc_dic[source][sector][file]= {}
        elif source == 'GEONAICS' and sector not in ['23','31','23']:
            disc_dic[source][sector]= {}
            for file in files:
                disc_dic[source][sector][file]= {}
```

```python
disc_dic['Basic']['23']['c'] = 'farts'
disc_dic['Basic']['23']['c'] 
```

```python
df = pd.DataFrame({'source': ['Basic', 'Basic', 'GEONAICS'],
                  'sector': ['23', '31','22'],
                  'type': ['c','c','d']})
df
```

```python
df['fun'] = df['source'].map(disc_dic)
```
