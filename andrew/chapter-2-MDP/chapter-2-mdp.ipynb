#### DTMPD: Discrete-Time MDP

**DTMPD** is a type of discret-time agent-enviroment interface discussed in chapter 1, remember how the agent and envirment interact with each other in a discrete-time agent-enviroment interface. The enviroment recieves an action and then agent recieves an observations durining interations. 

let the interaction time be $t = 0,1,2,...$

* the enviroiment generates rewards $R_t \in \mathcal{R}$   and reaches states $S_t \in \mathcal{S}$. where $\mathcal{R}$ is the **reward space** and $\mathcal{R} \subset \mathbb{R}$ and $\mathcal{S}$ is the **state space** which is  the set of all possible states/
* the agent obesres the enviroment and gets observation $O_t \in \mathcal{O}$ where $\mathcal{O}$  is in the **observation space** and is the set of all possible observations.  Then teh agente decides to conduct action $ A_t \in \mathcal{A}$, where $\mathcal{A}$ is the **action space**, the set of all possible actions

> _NOTE_ the candidates of states, observations, actions and rewards may differ for time $t$. For conveince of notations, we usually use a large space to cover all possible values, swe can use the same ntotation for all the steps.

a discrete-time agent-enviroment interace has the following **trajectory:**
$$R_0, S_0, O_0, R_1, S_1, O_1, R_2, S_2, O_2, ... $$

There is a terminal state for an episodeic tasks. The **terminal state** is denoted as $s_{end}$ and differse from other states: EPisode ends at the terminal state, and tehre will be no further observations and action rewards. Therefore the **terminal state**  $s_{end}$ is exclsed from state space $\mathcal{S}$ by default and we can use $\S^{+}$ to denote sthe state space wiht a terimal state

The trajectory of an episodic task is in the form of

$$R_0, S_0, O_0, R_1, S_1, O_1, R_2, S_2, O_2, ... , R_T, S_T = s_{end} $$

where $T$ is the number of steps to reach theend of the terminal state

> _NOTE_: the number of steps for an episode, denoted $T$ can be a random variable. From the view of a stochastic process, it can be viewd as a **stop time**

in the discrete-time agent-enviroment intervace if the agent can fully observe the state of the enviroment, we can assume $O_t = S_t (t= 0,1,2,...)$ without the loss of generality. In cushc cirumstances the trajectory can be simplifed to

$$R_0, S_0,  R_1, S_1, O_1, R_2, S_2, R_3, S_3, ... , R_T, S_T = s_{end} $$

the notation $O_t$ and $\mathcal{O}$ are no longer needed

> _NOTE_: the agent-enviroment interface does not require that the enivorment be fully observable, if the enviroment is not fully observable, the agent-enviroment interface can be modeled as other models such as **Partially observable Markov Decision Process (POMDP)** 





##### Stochastic process: Markov Pocess
for the time index set $\mathcal{T}$ (can be the set of natural numbers $\mathbb{N}$ or non-neative real numbers $[0, +\infty ]$). 

For a stochastic process $\{ S_T : t \in \mathcal{T} \}$. 

If for any $i \in \mathbb{N}, t_0,t_1...t_i, t_{i+1} \in \mathcal{T}$ and $s_0,s_1...s_i, s_{i+1} \in \mathcal{S}$  (we may assume $t_0, t_1...t_i, t_{i+1}$ without loss of generality). we have

$$Pr[S_{t_{i+1}} = S_{i+1}| S_{t_0} = S_0,  S_{t_1} = S_1 ... S_{t_i} = s_i] = Pr[S_{t_{i+1}} = s_{i+1}|S_{t_{i}} = s_{i} ]$$

we can say the process $\{S_t : t \geq 0 \}$ is a **Markov Process (MP)** or the process $\{S_t : t \geq 0 \}$ is **Markovian**  furhtermore, if for any $t, \tau \in \mathcal{T}$  and $s, s' \in \mathcal{S}$ we have $t + \tau \in \mathcal{T}$ and

$$Pr[S_{t+\tau} = s' | S_t = s] = Pr [S_{\tau} = s' | S_0 = s]$$


we say $\{S_t : t \geq 0 \}$ is homogenous MP. default assume homogenous MP.

given $\tau \in \mathcal{T}$ we can further define transition probablities over time interval $\tau$ as:

$$p^{[\tau]}(s'|s) \stackrel{\text{def}}{=} Pr[S_{\tau} = s' | S_0 =s], s,s' \in \mathcal{S}$$

It has the following properites:

$$p^{[0]}(s'|s) = 1_{[s' =s]}$$

where $1_{[.]}$ is the indicator function

$$p^{[\tau'+ \tau'']}(s'|s) = \sum_{s'' \in \mathcal{S}} p^{[\tau]} (s'|s'')p^{[\tau'']}p(s'',s), \tau', \tau'' \in \mathcal{T}, s',s'' \in \mathcal{S} $$

if we write $\tau$  transiption probabilty as a matrix $ \textbf{P}^{[\tau ]}$ in the shape of $|S| \times |S| $ where $|S|$ can be in infinite we get

$$ \textbf{P}^{[0]} = \textbf{I}$$
$$ \textbf{P}^{[\tau' =\tau'']} = \textbf{P}^{\tau'}\textbf{P}^{\tau''}, \tau', \tau'' \in \mathcal{T}$$

>_NOTE_ Deterministic materices are dnoted in bold upper case letters

If the time index set $\mathcal{T}$ is the sent of natural numbers $\mathbb{N}$ the MP is a **Discrete-Time Markov Process (DTMP)**. if If the time ndex $\mathcal{T}$ is $[0m + \inf)$ the MP is **Continious-Time Markov Process (CTMP)**






#### example DTMP

A **DTMP** has 2 states $\mathcal{S} = \{ \text{hungry}, \text{full} \}$ and its one step transtion probabilty is

$$

\begin{align}
& Pr[S_{t+1} = \text{hungry} |S_{t} = \text{hungry}] & = \frac{1}{2} \\ 
& Pr[S_{t+1} = \text{full} |S_{t} = \text{hungry}] & = \frac{1}{2} \\ 
& Pr[S_{t+1} = \text{hungry} |S_{t} = \text{full}] & = 0 \\ 
& Pr[S_{t+1} = \text{full} |S_{t} = \text{full}] & = 1\\ 
\end{align}

 $$

if a fully-observable discrete-time agent-enviroment interface satisfies that, at time $t$ the probablity to get reward $R_{t+1} = r$ and reach state $S_{t+1} = s'$ from $s_t =s$ and action $A_t = a$

$$Pr[S_{t+1} = s', R_{t+1} = r | S_t = s, A_t = a]$$

the discrete-time agent-enviroment interface becomes **Discrete-Time Markov Decision Process (DTMPD)**. This deinfition assumes that the reward $R_{t+ 1}$ and the next state $S_{t+1}$ only depend on the current state $S_t$ and current action $A_t$ and are independent of earlier states and actions (tl;dr Markovian see above...)
