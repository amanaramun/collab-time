#### DTMPD: Discrete-Time MDP

**DTMPD** is a type of discret-time agent-enviroment interface discussed in chapter 1, remember how the agent and envirment interact with each other in a discrete-time agent-enviroment interface. The enviroment recieves an action and then agent recieves an observations durining interations. 

let the interaction time be $t = 0,1,2,...$

* the enviroiment generates rewards $R_t \in \mathcal{R}$   and reaches states $S_t \in \mathcal{S}$. where $\mathcal{R}$ is the **reward space** and $\mathcal{R} \subset \mathbb{R}$ and $\mathcal{S}$ is the **state space** which is  the set of all possible states/
* the agent obesres the enviroment and gets observation $O_t \in \mathcal{O}$ where $\mathcal{O}$  is in the **observation space** and is the set of all possible observations.  Then teh agente decides to conduct action $ A_t \in \mathcal{A}$, where $\mathcal{A}$ is the **action space**, the set of all possible actions

> _NOTE_ the candidates of states, observations, actions and rewards may differ for time $t$. For conveince of notations, we usually use a large space to cover all possible values, swe can use the same ntotation for all the steps.

a discrete-time agent-enviroment interace has the following **trajectory:**
$$R_0, S_0, O_0, R_1, S_1, O_1, R_2, S_2, O_2, ... $$

There is a terminal state for an episodeic tasks. The **terminal state** is denoted as $s_{end}$ and differse from other states: EPisode ends at the terminal state, and tehre will be no further observations and action rewards. Therefore the **terminal state**  $s_{end}$ is exclsed from state space $\mathcal{S}$ by default and we can use $S^{+}$ to denote sthe state space wiht a terimal state


> _NOTE_ A "trajectory" is the sequence of what has happened (in terms of state, action, reward) over a set of contiguous timestamps, from a single episode, or a single part of a continuous problem. 

The trajectory of an episodic task is in the form of

$$R_0, S_0, O_0, R_1, S_1, O_1, R_2, S_2, O_2, ... , R_T, S_T = s_{end} $$

where $T$ is the number of steps to reach theend of the terminal state

> _NOTE_: the number of steps for an episode, denoted $T$ can be a random variable. From the view of a stochastic process, it can be viewd as a **stop time**

in the discrete-time agent-enviroment intervace if the agent can fully observe the state of the enviroment, we can assume $O_t = S_t (t= 0,1,2,...)$ without the loss of generality. In cushc cirumstances the trajectory can be simplifed to

$$R_0, S_0,  R_1, S_1, O_1, R_2, S_2, R_3, S_3, ... , R_T, S_T = s_{end} $$

the notation $O_t$ and $\mathcal{O}$ are no longer needed

> _NOTE_: the agent-enviroment interface does not require that the enivorment be fully observable, if the enviroment is not fully observable, the agent-enviroment interface can be modeled as other models such as **Partially observable Markov Decision Process (POMDP)** 





##### Stochastic process: Markov Pocess
for the time index set $\mathcal{T}$ (can be the set of natural numbers $\mathbb{N}$ or non-neative real numbers $[0, +\infty ]$). 

For a stochastic process $\{ S_T : t \in \mathcal{T} \}$. 

If for any $i \in \mathbb{N}, t_0,t_1...t_i, t_{i+1} \in \mathcal{T}$ and $s_0,s_1...s_i, s_{i+1} \in \mathcal{S}$  (we may assume $t_0, t_1...t_i, t_{i+1}$ without loss of generality). we have

$$Pr[S_{t_{i+1}} = S_{i+1}| S_{t_0} = S_0,  S_{t_1} = S_1 ... S_{t_i} = s_i] = Pr[S_{t_{i+1}} = s_{i+1}|S_{t_{i}} = s_{i} ]$$

we can say the process $\{S_t : t \geq 0 \}$ is a **Markov Process (MP)** or the process $\{S_t : t \geq 0 \}$ is **Markovian**  furhtermore, if for any $t, \tau \in \mathcal{T}$  and $s, s' \in \mathcal{S}$ we have $t + \tau \in \mathcal{T}$ and

$$Pr[S_{t+\tau} = s' | S_t = s] = Pr [S_{\tau} = s' | S_0 = s]$$


we say $\{S_t : t \geq 0 \}$ is homogenous MP. default assume homogenous MP.

given $\tau \in \mathcal{T}$ we can further define transition probablities over time interval $\tau$ as:

$$p^{[\tau]}(s'|s) \stackrel{\text{def}}{=} Pr[S_{\tau} = s' | S_0 =s], s,s' \in \mathcal{S}$$

It has the following properites:

$$p^{[0]}(s'|s) = 1_{[s' =s]}$$

where $1_{[.]}$ is the indicator function

$$p^{[\tau'+ \tau'']}(s'|s) = \sum_{s'' \in \mathcal{S}} p^{[\tau]} (s'|s'')p^{[\tau'']}p(s'',s), \tau', \tau'' \in \mathcal{T}, s',s'' \in \mathcal{S} $$

if we write $\tau$  transiption probabilty as a matrix $ \textbf{P}^{[\tau ]}$ in the shape of $|S| \times |S| $ where $|S|$ can be in infinite we get

$$ \textbf{P}^{[0]} = \textbf{I}$$
$$ \textbf{P}^{[\tau' =\tau'']} = \textbf{P}^{\tau'}\textbf{P}^{\tau''}, \tau', \tau'' \in \mathcal{T}$$

If the time index set $\mathcal{T}$ is the sent of natural numbers $\mathbb{N}$ the MP is a **Discrete-Time Markov Process (DTMP)**. if If the time ndex $\mathcal{T}$ is $[0m + \inf)$ the MP is **Continious-Time Markov Process (CTMP)**

 $$

if a fully-observable discrete-time agent-enviroment interface satisfies that, at time $t$ the probablity to get reward $R_{t+1} = r$ and reach state $S_{t+1} = s'$ from $s_t =s$ and action $A_t = a$

$$Pr[S_{t+1} = s', R_{t+1} = r | S_t = s, A_t = a]$$

the discrete-time agent-enviroment interface becomes **Discrete-Time Markov Decision Process (DTMPD)**. This deinfition assumes that the reward $R_{t+ 1}$ and the next state $S_{t+1}$ only depend on the current state $S_t$ and current action $A_t$ and are independent of earlier states and actions (tl;dr Markovian see above...)

>_note_ adding similar constraints on continious-time agent-enviroment inteface can lead to **Continious_time Markov Decision Process**


the main focus will be **homogenous Markov decision process** where the probabilty from state $S_t = s$  and  $A_t = a$ to reward $R_{t+1} =r$ and next state $ S_{t+1} = s'$

i.e.

$Pr[S_{t+1} = s' , R_{t+1} = r | S_t = s, A_t = a]$ does not change over $t$ and we can rewrite it as
$$
\begin{align*}
Pr[S_{t+1} = s' , R_{t+1} = r | S_t = s, A_t = a] &= Pr[S_{1} = s' , R_{1} = r | S_0 = s, A_0 = a]\\
& t \in \mathcal{T}, s \in \mathcal{S}, a \in \mathcal{A}, r \in \mathcal{R} , s' \in \mathcal{S}^+
\end{align*}
$$
> _NOTE_ **MPD** can either be homogenous or non-homogenous. non-homogenous will be discussed with continous-time later

Ina  **DTMPD** the first reward $R_0$ is out of the control of the agent. Additionally, after the enviroment provides the firs state $S_0$, all the following will be independent of $R_0$. Therefore we can exclude $R_0$ from teh system. The trajectory of **DTMPD** can be represented as

$$ S_0, A_0, R_1, S_1, A_1, R_2, S_2, A_2, R_3, ... S_t = s_{end}$$

Action and rewards are the important featuers of any **MDP**.

if we delete all the actions in a trajectory of an **MDP**  the trajectory becomes a **Markov Reward Process (MRP)**

$$ S_0, R_1, S_1, R_2, S_2, R_3, ... S_t = s_{end}$$

A **DTMPD** is a **finite MPD** if and only if  the state space $\mathcal{S}$ action space $\mathcal{A}$ and rewadr space $\mathcal{R}$ are all finite  

if we delete all rewards from a trajectoryu of an **MRP** it is now a trajectory of a **Markov Process (MP)**

####  Discounted Return

Consider an episodic task. Let $T$ denote the number of steps in an epsidoe. The **return** started from time $t$ denoted as $G_t$ can be deinfed as

$$ 
    \begin{align*}
        G_t \stackrel{def}{=} R_{t+1} +R_{t+2}+ ... +R_T \text{ episodict task, wihtout discount}
    \end{align*}
$$ 

> _note_ the number of steps $T$ can be a randon variable, therefore, in the definition of $G_T$ not only every term is ranodm, but als the number of terms is random

for sequential tasks, we also want to defin their returns o they can include all reards after time $t$. However, sequential tasks have no ending time, so the  aformetnioned denfition may not converage. In order to solve this problem a concept called `discont` is introduced, and disc soutned returns is defined as 

$$  G_t \stackrel{def}{=} \gamma^0 * R_{t+1} + \gamma^1 * R_{t+2}+ \gamma^2 + R_{t+3}+ \gamma^3 * R_{t+4}+ ... = \sum_{\tau = 0}^{+ \infty} \gamma^\tau R_{t+\tau+1}$$

where $\gamma \in [0,1]$ is called the **discount factor**. the discont factor trades off between current rewards and future retrns. Every unit of reward after $\tau$is eqlivalent to the $\gamma^\tau$ current rewards. in an exterme case, $\gamma = 0$ means the agent is myopic  and ignores future rewarda at all, and $\gamma = 1$  means that every future reard is as imporant as the curretn reward. Sequential tasks usually set $\gamma \in (0,1)$. in this case if the reard for every step is bonded the return is bonded.

To combine the two we can define $G_t$ as
$$ G_t \stackrel{def}{=} \sum_{\tau = 0}^{+\infty} \gamma^\tau R_{t+\tau+1}$$

for episodic tasks let $R_t = 0$ for $ t>T$.  In fact, the discont factor for an episodict tasks can be $<1$. Concequentyly, discont factor for an episodic tasks usually $\gamma \in (0,1]$, while discont factors for sequential tasks  are usually $\gamma \in (0,1)$

> _NOTE_ the preformance of *MDP* can  be other than metirc than the expectation of discoutned return. for example, the average reward can be used as a preformance metric.

discounted returns has the  following recurusive relationship:
$$G_t = R_{t+1} + \gamma G_{t+1}$$

this is due to 
$$ 
    \begin{align*}
        G_t = & \sum_{\tau =0}^{+\infty} \gamma^{\tau} R_{t+\tau +1} &\\
            = & R_{t+1} + \sum_{\tau =1}^{+\infty} \gamma^{\tau+1} R_{t+\tau +1} &\\
            = & R_{t+1} + \sum_{\tau =1}^{+\infty} \gamma^{\tau+1} R_{(t+1)+\tau +1} & \\
            = & R_{t+1} + \gamma^1 * \sum_{\tau =0}^{+\infty} \gamma^{\tau} R_{(t+1)+\tau +1} & \\
            = & R_{t+1} + \gamma* G_{t+1} &\text { $G_t \stackrel{def}{=} \sum_{\tau = 0}^{+\infty} \gamma^\tau R_{t+\tau+1}$ from above }
    \end{align*}
$$

This **discount return** of the whole trajectory is $G_0$. Its expectation is called the ** inital expected return** 
$$g_\pi \stackrel{def}{=}E_\pi[G_0]$$
 the agent always watns to maximize $g_\pi$  by choosing a smart policy. In order to emphasize the intetion to maximize the expectation of discoutned return, such **MDP** is sometimes called **discoutned MDP**

#### Value
Based on the defiti9on of return, we can further define the concept of value, which is a very important concept in RL. 

##### defintion of Value

With a fixed dynaimcis (dynamics refer to the rules that govern how the system transitions from one state to another based on the actions taken by the decision-maker. ) we can define **value** of policy $\pi$  as


* **state value** (a.k.a **discounted state value**) $v_{\pi}(s)$ is the expcted return starting from state $s$ if all subsequent actions are decided by the policy $\pi$

$$v_\pi(s) \stackrel{def}{=} E_\pi [G_t |S_t = s], s \in \mathcal{S}$$

* **action value** (a.k.a ** dsicouted cation value** ) $q_\pi(s,a)$ is the expected retunr starting from teh state-action pair $(s,a)$ if all subsequent atiosn are decided by the policy $\pi$


$$ q_\pi (s,a) \stackrel{def}{=} E_\pi [G_t|S_t = s, A_t = a], s \in \mathcal{S}, a \in \mathcal{A}$$

_Example :_
$$
    \begin{align*}
        & V_\pi(hungry)          & = &  E_\pi[G_t|S_t = hungry] \\
        & V_\pi(full)            & = &  E_\pi[G_t|S_t = full]  \\
        & q_\pi(hungry,feed)     & = & E_\pi[G_t|S_t = hungry, A_t = feed] \\
        & q_\pi(hungry,ignore)   & = & E_\pi[G_t|S_t = hungry, A_t = ignore] \\
        & q_\pi(full,feed)       & = & E_\pi[G_t|S_t = full, A_t = feed] \\
        & q_\pi(full,ignore)     & = & E_\pi[G_t|S_t = full, A_t = ignore] \\
    \end{align*}
$$

it is important toknow the value of a polcy. calcualting the values of  a polcy is called ** policy evaluation.** RL researchers have spent lots of effects  to find out good ways to evaluate a polcy.

#### Properties of value

This section intructoes some relationships between values, including the famous ** Bellman Expectation equations**.  these properties are useful for evaluating a policy.

lets cosnider the relationship between state values and action values.
State values can backup action values and action values can back up state values.

* USe action values to back up state values

$$ v_\pi (s) = \sum_{a \in \mathcal{A}} \pi(a|s)q_\pi(s,a), s \in \mathcal{S}$$

> _note_ in laymans termes the value of a state ($v_\pi (s)$) is the sum of policy (action $\pi(a|s)$) times return of the action $q_\pi(s,a)$. it is important to notice that the left hand side is in terms of State at time $t$, and the right hand side of the equation involves action  at time $t$. Therefore the relationship uses action values at time to to back up state values at time $t$ and can be written as

$$ v_\pi (S_t) = E_\pi [q_\pi(S_t,A_t)]$$

this relationship can be illustarted by a **backup diagram** where hallow circles reprsent the present state, and solid circles represent the state-action pairs.
